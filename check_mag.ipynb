```python
import numpy as np
from pathlib import Path
from glob import glob


def read_last_n_numeric_columns(filename, ncols):
    rows = []

    with open(filename, "r") as f:
        for line in f:
            line = line.strip()

            if not line or line.startswith("#"):
                continue

            parts = line.split()

            try:
                nums = [float(x) for x in parts]
            except ValueError:
                continue

            if len(nums) >= ncols:
                rows.append(nums[-ncols:])

    if not rows:
        raise ValueError(f"No valid numeric rows found in {filename}")

    return np.array(rows, dtype=np.float32)


def build_spin_config_array(
    position_file,
    spin_pattern="spins-*.data",
):
    """
    Returns:
        configs.shape = (n_files, n_atoms, 6)

    Columns:
        x y z Sx Sy Sz
    """

    # Read positions once
    positions = read_last_n_numeric_columns(position_file, 3)

    # Find all spin files
    spin_files = sorted(glob(spin_pattern))

    if len(spin_files) == 0:
        raise ValueError(f"No spin files found matching: {spin_pattern}")

    configs = []

    for spin_file in spin_files:

        spins = read_last_n_numeric_columns(spin_file, 3)

        if spins.shape[0] != positions.shape[0]:
            raise ValueError(
                f"Mismatch in {spin_file}:\n"
                f"positions rows = {positions.shape[0]}\n"
                f"spin rows      = {spins.shape[0]}"
            )

        # shape = (n_atoms, 6)
        config = np.column_stack((positions, spins))

        configs.append(config)

        print(f"Loaded {spin_file}")

    # shape = (n_files, n_atoms, 6)
    configs = np.array(configs, dtype=np.float32)

    print("\nFinal array shape:", configs.shape)

    return configs, spin_files


# Example usage
configs, spin_files = build_spin_config_array(
    position_file="atoms-coords.data",
    spin_pattern="spins-*.data",
)

print(configs.shape)
```

```python
import numpy as np
from scipy.spatial import cKDTree


def build_neighbor_pairs_from_positions(
    positions,
    cutoff,
):
    """
    Finds pairs i < j with distance < cutoff.

    positions.shape = (n_atoms, 3)
    """

    tree = cKDTree(positions)
    pairs = tree.query_pairs(r=cutoff)

    pairs = np.array(sorted(pairs), dtype=int)

    return pairs


def compute_energies_heisenberg(
    configs,
    pairs,
    J=1.0,
):
    """
    Computes one energy per spin configuration.

    configs.shape = (n_configs, n_atoms, 6)
    pairs.shape   = (n_pairs, 2)

    Hamiltonian:
        H = -J sum_<ij> S_i · S_j
    """

    spins = configs[:, :, 3:6]

    i = pairs[:, 0]
    j = pairs[:, 1]

    Si = spins[:, i, :]
    Sj = spins[:, j, :]

    dot_products = np.sum(Si * Sj, axis=2)

    energies = -J * np.sum(dot_products, axis=1)

    return energies

import numpy as np
import numpy as np
from scipy.spatial import cKDTree


def build_neighbor_pairs_from_positions(positions, cutoff):
    tree = cKDTree(positions)
    return np.array(sorted(tree.query_pairs(r=cutoff)), dtype=int)


def check_quantized_conditional_distribution_many(
    configs,
    pairs,
    J=1.0,
    S=1.0,
    temperature=325.0,
    mu_s_SI=1.0,
    alpha=1.0,
    Tc=660.0,
    n_bins_B=10,
    start=500,
):
    """
    configs: (n_configs, n_atoms, 6), columns 3:6 are spins
    pairs:   (n_pairs, 2)

    Uses the same beta_eff convention as your C++ heat bath:

        kBTBohr = mu_B / (T_rescaled k_B)
        beta_eff = mu_s_SI * (1/mu_B) * kBTBohr

    where 1/mu_B = 1.07828231e23.
    """

    pairs = np.asarray(pairs, dtype=int)

    spins_all = np.asarray(configs[start:, :, 3:6], dtype=float)
    n_cfg, n_atoms, _ = spins_all.shape

    i = pairs[:, 0]
    j = pairs[:, 1]

    twoS = int(round(2 * S))
    m_vals = np.array([-S + p for p in range(twoS + 1)], dtype=float)
    u_allowed = m_vals / S

    # VAMPIRE temperature rescaling convention
    if temperature < Tc:
        T_rescaled = Tc * (temperature / Tc) ** alpha
    else:
        T_rescaled = temperature

    kBTBohr = 9.27400915e-24 / (T_rescaled * 1.3806503e-23)
    beta_eff = mu_s_SI * 1.07828231e23 * kBTBohr

    all_Bnorm = []
    all_idx_obs = []
    all_quant_error = []

    for spins in spins_all:
        B = np.zeros_like(spins)

        # local field B_i = J sum_j S_j
        np.add.at(B, i, J * spins[j])
        np.add.at(B, j, J * spins[i])

        Bnorm = np.linalg.norm(B, axis=1)
        mask = Bnorm > 1e-14

        Bhat = np.zeros_like(B)
        Bhat[mask] = B[mask] / Bnorm[mask, None]

        u = np.sum(spins * Bhat, axis=1)

        idx_obs = np.argmin(np.abs(u[:, None] - u_allowed[None, :]), axis=1)
        quant_error = np.abs(u - u_allowed[idx_obs])

        all_Bnorm.append(Bnorm[mask])
        all_idx_obs.append(idx_obs[mask])
        all_quant_error.append(quant_error[mask])

    B_used = np.concatenate(all_Bnorm)
    idx_used = np.concatenate(all_idx_obs)
    quant_error = np.concatenate(all_quant_error)

    print("Number of pooled site samples:", len(B_used))
    print("T =", temperature)
    print("T_rescaled =", T_rescaled)
    print("beta_eff =", beta_eff)
    print("Mean quantization error:", np.mean(quant_error))
    print("Max quantization error :", np.max(quant_error))

    edges = np.linspace(B_used.min(), B_used.max(), n_bins_B + 1)

    print("\nConditional distribution test:")
    print("m values:", m_vals)
    print("bin   B_mean      count   observed probs                 predicted probs")

    results = []

    for b in range(n_bins_B):
        if b == n_bins_B - 1:
            in_bin = (B_used >= edges[b]) & (B_used <= edges[b + 1])
        else:
            in_bin = (B_used >= edges[b]) & (B_used < edges[b + 1])

        count = int(np.sum(in_bin))
        if count < 50:
            continue

        B_mean = np.mean(B_used[in_bin])

        counts = np.array(
            [np.sum(idx_used[in_bin] == p) for p in range(len(m_vals))],
            dtype=float,
        )

        p_obs = counts / counts.sum()

        args = beta_eff * B_mean * m_vals
        args -= np.max(args)
        weights = np.exp(args)
        p_pred = weights / weights.sum()

        print(
            f"{b:2d}  {B_mean:10.5g}  {count:7d}   "
            f"{p_obs}   {p_pred}"
        )

        results.append((B_mean, count, p_obs, p_pred))

    return results


positions = configs[0, :, 0:3]

pairs = build_neighbor_pairs_from_positions(
    positions,
    cutoff=2.5,
)

results = check_quantized_conditional_distribution_many(
    configs=configs,
    pairs=pairs,
    J=1.0,
    S=1.0,
    temperature=325.0,
    mu_s_SI=1.0,     # replace by your material mu_s_SI if known
    alpha=1.0,       # replace by VAMPIRE material alpha
    Tc=660.0,
    n_bins_B=10,
    start=500,
)
```

```python
def fit_beta_eff_from_results(results, eps=1e-12):
    xs = []
    ys = []

    for B_mean, count, p_obs, p_pred_dummy in results:
        p_minus, p_zero, p_plus = p_obs

        if p_zero > eps and p_plus > eps:
            xs.append(B_mean)
            ys.append(np.log(p_plus / p_zero))

    xs = np.array(xs)
    ys = np.array(ys)

    beta_eff_fit = np.sum(xs * ys) / np.sum(xs * xs)
    return beta_eff_fit


def print_conditional_with_fitted_beta(results, S=1.0):
    m_vals = np.array([-S, 0.0, S])

    beta_eff_fit = fit_beta_eff_from_results(results)

    print("Fitted beta_eff =", beta_eff_fit)
    print("\nB_mean      observed probs                 fitted predicted probs")

    for B_mean, count, p_obs, _ in results:
        args = beta_eff_fit * B_mean * m_vals
        args -= np.max(args)
        w = np.exp(args)
        p_fit = w / w.sum()

        print(f"{B_mean:10.5g}   {p_obs}   {p_fit}")

    return beta_eff_fit

beta_eff_fit = print_conditional_with_fitted_beta(results, S=1.0)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def load_energy_trace(filename, discard_fraction=0.5):
    data = np.loadtxt(filename, comments="#")
    sweep = data[:, 0]
    T = data[:, 1]
    E = data[:, 2]

    n0 = int(discard_fraction * len(E))
    return sweep[n0:], T[n0:], E[n0:]


def detailed_balance_current_test(
    filename,
    n_bins=30,
    discard_fraction=0.5,
    min_pair_count=10,
):
    sweep, T, E = load_energy_trace(filename, discard_fraction)

    # Energy bins
    edges = np.linspace(E.min(), E.max(), n_bins + 1)
    states = np.digitize(E, edges) - 1
    states = np.clip(states, 0, n_bins - 1)

    # Transition count matrix
    C = np.zeros((n_bins, n_bins), dtype=np.int64)

    for a, b in zip(states[:-1], states[1:]):
        C[a, b] += 1

    # Antisymmetric current
    J = C - C.T
    pair_total = C + C.T

    # Normalized current per pair
    valid = pair_total >= min_pair_count
    np.fill_diagonal(valid, False)

    J_norm = np.zeros_like(J, dtype=float)
    J_norm[valid] = J[valid] / pair_total[valid]

    rms_current = np.sqrt(np.mean(J_norm[valid] ** 2)) if np.any(valid) else np.nan
    max_current = np.max(np.abs(J_norm[valid])) if np.any(valid) else np.nan

    print("File:", filename)
    print("Samples after discard:", len(E))
    print("T mean:", np.mean(T))
    print("E mean:", np.mean(E))
    print("E std:", np.std(E, ddof=1))
    print("Number of valid bin pairs:", np.sum(valid) // 2)
    print("RMS normalized current:", rms_current)
    print("Max normalized current:", max_current)

    # Plot energy histogram
    plt.figure(figsize=(6, 4))
    plt.hist(E, bins=edges, density=True, histtype="step", linewidth=2)
    plt.xlabel("Energy")
    plt.ylabel("P(E)")
    plt.tight_layout()
    plt.show()

    # Plot transition matrix
    plt.figure(figsize=(5, 4))
    plt.imshow(C, origin="lower", aspect="auto")
    plt.colorbar(label="N(a→b)")
    plt.xlabel("new energy bin")
    plt.ylabel("old energy bin")
    plt.title("Transition counts")
    plt.tight_layout()
    plt.show()

    # Plot antisymmetric current
    plt.figure(figsize=(5, 4))
    plt.imshow(J_norm, origin="lower", aspect="auto", vmin=-1, vmax=1)
    plt.colorbar(label="(N_ab - N_ba)/(N_ab + N_ba)")
    plt.xlabel("b")
    plt.ylabel("a")
    plt.title("Normalized probability current")
    plt.tight_layout()
    plt.show()

    return {
        "E": E,
        "T": T,
        "states": states,
        "edges": edges,
        "C": C,
        "J": J,
        "J_norm": J_norm,
        "valid": valid,
        "rms_current": rms_current,
        "max_current": max_current,
    }

res = detailed_balance_current_test(

    "mc_energy_trace.dat",

    n_bins=20,

    discard_fraction=0.5,

    min_pair_count=50,

)
```
