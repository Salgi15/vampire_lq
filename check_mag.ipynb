```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve, curve_fit

arr2 = np.array([
    [0.2,   1.0],
    [25.2,  1.0],
    [50.2,  1.0],
    [75.2,  0.999836],
    [100.2, 0.999342],
    [125.2, 0.999506],
    [150.2, 0.998352],
    [175.2, 0.994737],
    [200.2, 0.989134],
    [225.2, 0.986664],
    [250.2, 0.976479],
    [275.2, 0.96611],
    [300.2, 0.954198],
    [325.2, 0.944856],
    [350.2, 0.920035],
    [375.2, 0.919869],
    [400.2, 0.878157],
    [425.2, 0.856729],
    [450.2, 0.82088],
    [475.2, 0.777818],
    [500.2, 0.749838],
    [525.2, 0.667091],
    [550.2, 0.633441],
    [575.2, 0.584558],
    [600.2, 0.489411],
    [625.2, 0.337027],
    [650.2, 0.252173],
    [675.2, 0.0689342],
    [700.2, 0.0417088],
    [725.2, 0.0612186],
    [750.2, 0.0229967],
    [775.2, 0.0680477],
    [800.2, 0.059653],
    [825.2, 0.0399003],
    [850.2, 0.0286409],
    [875.2, 0.0272205],
    [900.2, 0.0148732],
    [925.2, 0.0101311],
    [950.2, 0.0255184],
])

T = arr[:, 0]
m_data = arr[:, 1]

def B1(x):
    x = np.asarray(x, dtype=float)
    out = np.empty_like(x)
    small = np.abs(x) < 1e-8
    out[small] = (2.0 / 3.0) * x[small]
    xs = x[~small]
    out[~small] = 1.5 / np.tanh(1.5 * xs) - 0.5 / np.tanh(0.5 * xs)
    return out

def m_brillouin_S1(T, Tc):
    T = np.asarray(T, dtype=float)
    m = np.zeros_like(T)

    for i, Ti in enumerate(T):
        if Ti >= Tc:
            m[i] = 0.0
        else:
            def f(mm):
                return mm - B1(np.array([1.5 * (Tc / Ti) * mm]))[0]

            guess = 0.999 if Ti < 0.8 * Tc else 0.5
            m[i] = fsolve(f, guess)[0]

    return m

def fitfun(T, Tc):
    return m_brillouin_S1(T, Tc)

# fit Tc to arr2
popt, _ = curve_fit(fitfun, T, m_data, p0=[680], bounds=(1.0, 2000.0))
Tc_fit = popt[0]

Tfine = np.linspace(T.min(), T.max(), 800)
m_fit = m_brillouin_S1(Tfine, Tc_fit)

plt.figure(figsize=(7,5))
plt.plot(T, m_data, 'o', label='SMC data')
plt.plot(Tfine, m_fit, '-', label=fr'S=1 Brillouin fit, $T_c={Tc_fit:.1f}$ K')
plt.xlabel('T (K)')
plt.ylabel('m')
plt.legend()
plt.tight_layout()
plt.show()

print(f"Fitted Tc = {Tc_fit:.3f} K")
```
